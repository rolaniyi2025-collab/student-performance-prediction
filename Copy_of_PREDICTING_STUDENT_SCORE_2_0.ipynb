## 4.FEATURE SELECTION, SPLITING AND SCALING

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# DEFINE FEATURES AND TARGET

X = df[['Age', 'Gender', 'Ethnicity', 'ParentalEducation',
        'StudyTimeWeekly', 'Absences', 'Tutoring',
        'ParentalSupport', 'Extracurricular',
        'Sports', 'Music', 'Volunteering']].copy()

# One-hot encode Ethnicity because it represents categories
X = pd.get_dummies(X, columns=['Ethnicity'], dtype=int)

y_gpa = df['GPA']               # GPA target for regression
y_class = df['GradeClass']      # GradeClass target for classification later
```

```python
# Split before scaling to avoid data leakage
X_train, X_test, y_gpa_train, y_gpa_test, y_class_train, y_class_test = train_test_split(
    X,
    y_gpa,
    y_class,
    test_size=0.2,
    random_state=42,
    stratify=y_class
)
```

```python
# Scale features using StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) # Fit only on training data
X_test_scaled = scaler.transform(X_test) # Transform test using train's stats
```

## 5.LINEAR REGRESSION MODEL

```python
# Train the GPA regression model
gpa_model = LinearRegression()
gpa_model.fit(X_train_scaled, y_gpa_train)

# Predict GPA for the test students
y_pred_gpa = gpa_model.predict(X_test_scaled)
```

```python
# 5-Fold Cross-Validation

gpa_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LinearRegression())
])

cv = KFold(n_splits=5, shuffle=True, random_state=42)

# Cross-validation R² scores
cv_r2 = cross_val_score(
    gpa_pipeline,
    X,
    y_gpa,
    cv=cv,
    scoring='r2'
)

# Cross-validation RMSE scores
cv_rmse = -cross_val_score(
    gpa_pipeline,
    X,
    y_gpa,
    cv=cv,
    scoring='neg_root_mean_squared_error'
)

print("5-Fold Cross-Validation Results")
print("--------------------------------")
print("R² scores:", cv_r2)
print("Mean R²:", cv_r2.mean())
print("Std R²:", cv_r2.std())

print("\nRMSE scores:", cv_rmse)
print("Mean RMSE:", cv_rmse.mean())
print("Std RMSE:", cv_rmse.std())
```

## 6.MODEL EVALUATION

```python
import numpy as np

mae = mean_absolute_error(y_gpa_test, y_pred_gpa)
mse = mean_squared_error(y_gpa_test, y_pred_gpa)
rmse = np.sqrt(mse)
r2 = r2_score(y_gpa_test, y_pred_gpa)

print(f"MAE: {mae:.3f}")
print(f"MSE: {mse:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R²: {r2:.4f}")
```

## 7.ACTUAL VS PREDICTED SCORES AND PLOT

```python
def gpa_to_gradeclass(gpa):
    if gpa >= 3.5:
        return 0
    elif gpa >= 3.0:
        return 1
    elif gpa >= 2.5:
        return 2
    elif gpa >= 2.0:
        return 3
    else:
        return 4

hybrid_pred_class = [gpa_to_gradeclass(gpa) for gpa in y_pred_gpa]
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    confusion_matrix,
    classification_report
)

hybrid_accuracy = accuracy_score(y_class_test, hybrid_pred_class)
hybrid_f1 = f1_score(
    y_class_test,
    hybrid_pred_class,
    average='weighted'
)

print(f"Hybrid Accuracy: {hybrid_accuracy:.4f}")
print(f"Hybrid Weighted F1-score: {hybrid_f1:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_class_test, hybrid_pred_class))

print("\nClassification Report:")
print(classification_report(
    y_class_test,
    hybrid_pred_class,
    zero_division=0
))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Directly predict GradeClass from student features
class_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42
)

class_model.fit(X_train, y_class_train)

direct_pred_class = class_model.predict(X_test)
```

```python
direct_accuracy = accuracy_score(y_class_test, direct_pred_class)
direct_f1 = f1_score(y_class_test, direct_pred_class, average='weighted')

print(f"Direct Classification Accuracy: {direct_accuracy:.4f}")
print(f"Direct Classification Weighted F1-score: {direct_f1:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_class_test, direct_pred_class))
```

```python
direct_macro_f1 = f1_score(
    y_class_test,
    direct_pred_class,
    average='macro'
)

print(f"Direct Classification Macro F1-score: {direct_macro_f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_class_test,
    direct_pred_class,
    zero_division=0
))
```

```python
hybrid_macro_f1 = f1_score(
    y_class_test,
    hybrid_pred_class,
    average='macro'
)

comparison = pd.DataFrame({
    'Method': [
        'Hybrid: GPA → GradeClass',
        'Direct Classification'
    ],
    'Accuracy': [
        hybrid_accuracy,
        direct_accuracy
    ],
    'Weighted F1-score': [
        hybrid_f1,
        direct_f1
    ],
    'Macro F1-score': [
        hybrid_macro_f1,
        direct_macro_f1
    ]
})

comparison
```

MODELLING APPROACHES

1. HYBRID REGRESSION + CLASSIFICATION
- Linear Regression predicts continuous GPA.
- The predicted GPA is converted to GradeClass using the dataset-defined boundaries:
  GPA >= 3.5 → Class 0
  GPA >= 3.0 → Class 1
  GPA >= 2.5 → Class 2
  GPA >= 2.0 → Class 3

- Random Forest directly predicts GradeClass.

PREPROCESSING
- No missing values.
- No duplicate records.
- StudentID is excluded from the features.
- Ethnicity is one-hot encoded.
- Train/test split is 80/20 and stratified by GradeClass.
- StandardScaler is fitted only on the training data and applied to the test data.
- Cross-validation was considered but NOT added because the project should remain reasonably simple.
